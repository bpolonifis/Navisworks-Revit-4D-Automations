```python

import uuid
# -----------------------------------
# GUID Generator
# -----------------------------------
def generate_guid():
    return str(uuid.uuid4())

# -----------------------------------
# XML Header
# -----------------------------------
output = f"""<?xml version="1.0" encoding="UTF-8" ?>
<exchange xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance"
xsi:noNamespaceSchemaLocation="http://download.autodesk.com/us/navisworks/schemas/nw-exchange-12.0.xsd"
units="ft" filename="" filepath="">
  <selectionsets>
    <viewfolder name="WORKSETS_LEVELS" guid="{generate_guid()}">
"""

# -----------------------------------
# Selectionset Template (Workset × Layer>Level)
# -----------------------------------
selectionset_template = """      <selectionset name="{name}" guid="{guid}">
        <findspec mode="all" disjoint="1">
          <conditions>
            <condition test="contains" flags="0">
              <category>
                <name internal="LcRevitData_Element">Element</name>
              </category>
              <property>
                <name internal="lcldrevit_parameter_-1002053">Workset</name>
              </property>
              <value>
                <data type="wstring">{workset}</data>
              </value>
            </condition>
            <condition test="contains" flags="0">
              <category>
                <name internal="LcOaNode">Item</name>
              </category>
              <property>
                <name internal="LcOaNodeLayer">Layer</name>
              </property>
              <value>
                <data type="wstring">{level}</data>
              </value>
            </condition>
          </conditions>
          <locator>/</locator>
        </findspec>
      </selectionset>
"""

# -----------------------------------
# Generate Workset × Level combinations
# -----------------------------------
for ws in worksets:
    for lvl in levels:
        selection_name = f"{ws} | {lvl}"
        output += selectionset_template.format(
            name=selection_name,
            guid=generate_guid(),
            workset=ws,
            level=lvl
        )

# -----------------------------------
# XML Footer
# -----------------------------------
output += """    </viewfolder>
  </selectionsets>
</exchange>"""

# -----------------------------------
# Print XML
# -----------------------------------
#print(output)


# -----------------------------------
# Export XML to file
# -----------------------------------
output_path = "/content/drive/MyDrive/WORKSETS_LEVELS.xml"   # Change path if needed

with open(output_path, "w", encoding="utf-8") as f:
    f.write(output)

print(f"XML successfully exported to: {output_path}")
```

# IFC Handling

```python
import ifcopenshell
import pandas as pd
from google.colab import files

# Upload IFC file
uploaded = files.upload()

# Get uploaded file name
ifc_filename = list(uploaded.keys())[0]

# Load IFC file
ifc_file = ifcopenshell.open(ifc_filename)

#########################################
# 1️⃣ Extract All IFC Entities
#########################################

entities_data = []

for entity in ifc_file:
    entity_dict = {}

    entity_dict["ExpressID"] = entity.id()
    entity_dict["GlobalId"] = getattr(entity, "GlobalId", None)
    entity_dict["EntityType"] = entity.is_a()

    # Get all attributes
    for attr_name, attr_value in zip(entity.get_info().keys(), entity.get_info().values()):
        entity_dict[attr_name] = str(attr_value)

    entities_data.append(entity_dict)

entities_df = pd.DataFrame(entities_data)

#########################################
# 2️⃣ Extract All Property Sets
#########################################

pset_data = []

for entity in ifc_file.by_type("IfcObject"):
    if hasattr(entity, "IsDefinedBy"):
        for definition in entity.IsDefinedBy:
            if definition.is_a("IfcRelDefinesByProperties"):
                prop_set = definition.RelatingPropertyDefinition

                if prop_set.is_a("IfcPropertySet"):
                    for prop in prop_set.HasProperties:
                        pset_dict = {}
                        pset_dict["Element_GlobalId"] = entity.GlobalId
                        pset_dict["Element_Type"] = entity.is_a()
                        pset_dict["Pset_Name"] = prop_set.Name
                        pset_dict["Property_Name"] = prop.Name

                        # Fix: Add check for prop.NominalValue being None
                        if hasattr(prop, "NominalValue") and prop.NominalValue is not None:
                            pset_dict["Property_Value"] = str(prop.NominalValue.wrappedValue)
                        else:
                            pset_dict["Property_Value"] = None

                        pset_data.append(pset_dict)

pset_df = pd.DataFrame(pset_data)

#########################################
# 3️⃣ Export to Excel
#########################################

output_filename = "IFC_Export.xlsx"

with pd.ExcelWriter(output_filename, engine="openpyxl") as writer:
    entities_df.to_excel(writer, sheet_name="All_Entities", index=False)
    pset_df.to_excel(writer, sheet_name="PropertySets", index=False)

print("Export completed!")

# Download file
files.download(output_filename)
```

#✅ Google Colab Script — Optimized for Tekla 2022 IFC

```python
import ifcopenshell
import pandas as pd
from google.colab import files

# Upload IFC
uploaded = files.upload()
ifc_filename = list(uploaded.keys())[0]

# Open IFC
ifc_file = ifcopenshell.open(ifc_filename)

############################################################
# FUNCTION: Extract Property Sets + Quantities (Tekla Safe)
############################################################

def get_psets_and_quantities(element):
    data = []

    if hasattr(element, "IsDefinedBy"):
        for rel in element.IsDefinedBy:
            if rel.is_a("IfcRelDefinesByProperties"):
                prop_def = rel.RelatingPropertyDefinition

                # Property Sets
                if prop_def.is_a("IfcPropertySet"):
                    for prop in prop_def.HasProperties:
                        value = None
                        if hasattr(prop, "NominalValue") and prop.NominalValue:
                            value = prop.NominalValue.wrappedValue

                        data.append({
                            "GlobalId": element.GlobalId,
                            "EntityType": element.is_a(),
                            "Pset_or_QuantitySet": prop_def.Name,
                            "Property_Name": prop.Name,
                            "Value": value
                        })

                # Quantities (BaseQuantities in Tekla)
                if prop_def.is_a("IfcElementQuantity"):
                    for quantity in prop_def.Quantities:
                        val = None

                        if hasattr(quantity, "LengthValue"):
                            val = quantity.LengthValue
                        elif hasattr(quantity, "AreaValue"):
                            val = quantity.AreaValue
                        elif hasattr(quantity, "VolumeValue"):
                            val = quantity.VolumeValue
                        elif hasattr(quantity, "CountValue"):
                            val = quantity.CountValue
                        elif hasattr(quantity, "WeightValue"):
                            val = quantity.WeightValue

                        data.append({
                            "GlobalId": element.GlobalId,
                            "EntityType": element.is_a(),
                            "Pset_or_QuantitySet": prop_def.Name,
                            "Property_Name": quantity.Name,
                            "Value": val
                        })

    return data

############################################################
# 1️⃣ Extract All Elements (Physical Products Only)
############################################################

elements_data = []
psets_data = []

for element in ifc_file.by_type("IfcProduct"):

    base_info = {
        "ExpressID": element.id(),
        "GlobalId": getattr(element, "GlobalId", None),
        "EntityType": element.is_a(),
        "Name": getattr(element, "Name", None),
        "ObjectType": getattr(element, "ObjectType", None)
    }

    elements_data.append(base_info)

    # Extract Tekla Psets & Quantities
    psets_data.extend(get_psets_and_quantities(element))

elements_df = pd.DataFrame(elements_data)
psets_df = pd.DataFrame(psets_data)

############################################################
# 2️⃣ Export to Excel
############################################################

output_filename = "Tekla_IFC_Export.xlsx"

with pd.ExcelWriter(output_filename, engine="openpyxl") as writer:
    elements_df.to_excel(writer, sheet_name="Elements", index=False)
    psets_df.to_excel(writer, sheet_name="Properties_Quantities", index=False)

print("Export Completed Successfully")

files.download(output_filename)
```

#✅ Step 3 — Merge Script

```python
import ifcopenshell
import ifcopenshell.guid
import os
import glob
import gc

# --------------------------------------------------
# USER INPUT
# --------------------------------------------------
input_folder = "/content/drive/MyDrive/MSP(IFC)"
output_folder = os.path.dirname(input_folder)
output_file = os.path.join(output_folder, "Merged_Model.ifc")

# --------------------------------------------------
# GET IFC FILES
# --------------------------------------------------
ifc_files = sorted(glob.glob(os.path.join(input_folder, "*.ifc")))

if not ifc_files:
    raise Exception("No IFC files found in folder!")

print(f"Found {len(ifc_files)} IFC files.")

# --------------------------------------------------
# OPEN FIRST FILE AS BASE MODEL
# --------------------------------------------------
print(f"Opening base model: {os.path.basename(ifc_files[0])}")
merged_model = ifcopenshell.open(ifc_files[0])

# --------------------------------------------------
# FUNCTION: SAFE ADD PRODUCT ONLY
# --------------------------------------------------
def safe_add_products(source_model, target_model, file_name):

    products = source_model.by_type("IfcProduct")

    print(f"  → Found {len(products)} products")

    for i, product in enumerate(products):

        try:
            # Regenerate GUID before adding
            if hasattr(product, "GlobalId"):
                product.GlobalId = ifcopenshell.guid.new()

            target_model.add(product)

            if i % 500 == 0:
                print(f"     Added {i} elements...")

        except Exception as e:
            print(f"     Skipped element ({product.is_a()}): {e}")

    print(f"  ✓ Finished merging {file_name}")


# --------------------------------------------------
# MERGE REMAINING FILES (SAFE MODE)
# --------------------------------------------------
for file_path in ifc_files[1:]:

    print(f"\nMerging: {os.path.basename(file_path)}")

    try:
        current_model = ifcopenshell.open(file_path)

        safe_add_products(current_model, merged_model, os.path.basename(file_path))

        # Free memory aggressively
        del current_model
        gc.collect()

    except Exception as e:
        print(f"Failed to process {file_path}: {e}")

# --------------------------------------------------
# WRITE MERGED FILE
# --------------------------------------------------
print("\nWriting merged IFC...")
merged_model.write(output_file)

print("✅ Merged IFC saved to:")
print(output_file)
```

```python
import os
import glob
import re

# --------------------------------------------------
# USER INPUT
# --------------------------------------------------
input_folder = "/content/drive/MyDrive/MSP(IFC)"
output_folder = os.path.dirname(input_folder)
output_file = os.path.join(output_folder, "Merged_Model.ifc")

ifc_files = sorted(glob.glob(os.path.join(input_folder, "*.ifc")))

if not ifc_files:
    raise Exception("No IFC files found!")

print(f"Found {len(ifc_files)} IFC files.")

# --------------------------------------------------
# HELPER: Extract HEADER + DATA
# --------------------------------------------------
def split_ifc(file_path):
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        content = f.read()

    header = content.split("DATA;")[0]
    data_section = content.split("DATA;")[1].split("ENDSEC;")[0]

    return header, data_section


# --------------------------------------------------
# START MERGING
# --------------------------------------------------

global_id_offset = 0
merged_data = []

print("Reading base model...")

base_header, base_data = split_ifc(ifc_files[0])
merged_data.append(base_data)

# Find max entity number in base file
matches = re.findall(r"#(\d+)", base_data)
if matches:
    global_id_offset = max(map(int, matches))

print(f"Starting entity offset: {global_id_offset}")

# --------------------------------------------------
# PROCESS REMAINING FILES
# --------------------------------------------------

for file_path in ifc_files[1:]:

    print(f"Merging {os.path.basename(file_path)}")

    _, data = split_ifc(file_path)

    def renumber(match):
        return f"#{int(match.group(1)) + global_id_offset}"

    renumbered_data = re.sub(r"#(\d+)", renumber, data)

    merged_data.append(renumbered_data)

    matches = re.findall(r"#(\d+)", renumbered_data)
    if matches:
        global_id_offset = max(map(int, matches))

# --------------------------------------------------
# WRITE FINAL IFC
# --------------------------------------------------

with open(output_file, "w", encoding="utf-8") as f:
    f.write(base_header)
    f.write("DATA;\n")

    for section in merged_data:
        f.write(section.strip() + "\n")

    f.write("ENDSEC;\nEND-ISO-10303-21;")

print("✅ Merge completed successfully.")
print("Saved to:", output_file)
```

```python
import uuid
import os
from openpyxl import load_workbook

# -----------------------------------
# Ask user for Excel file path
# -----------------------------------
excel_path = input("Enter full path to Excel file (.xlsx): ").strip()

if not os.path.exists(excel_path):
    print("File not found.")
    exit()

# -----------------------------------
# Create XML output path (same name, .xml)
# -----------------------------------
base_name = os.path.splitext(excel_path)[0]
xml_output_path = base_name + ".xml"

# -----------------------------------
# Load workbook
# -----------------------------------
wb = load_workbook(excel_path)

if "Worksets" not in wb.sheetnames or "Levels" not in wb.sheetnames:
    print("Sheets 'Worksets' and/or 'Levels' not found.")
    exit()

ws_sheet = wb["Worksets"]
lvl_sheet = wb["Levels"]

# Read data (skip header)
worksets = [r[0].value for r in ws_sheet.iter_rows(min_row=2) if r[0].value]
levels = [r[0].value for r in lvl_sheet.iter_rows(min_row=2) if r[0].value]

# -----------------------------------
# GUID Generator
# -----------------------------------
def generate_guid():
    return str(uuid.uuid4())

# -----------------------------------
# XML Header
# -----------------------------------
output = f"""<?xml version="1.0" encoding="UTF-8" ?>
<exchange xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance"
xsi:noNamespaceSchemaLocation="http://download.autodesk.com/us/navisworks/schemas/nw-exchange-12.0.xsd"
units="ft" filename="" filepath="">
  <selectionsets>
    <viewfolder name="WORKSETS_LEVELS" guid="{generate_guid()}">
"""

# -----------------------------------
# Selectionset Template
# -----------------------------------
selectionset_template = """      <selectionset name="{name}" guid="{guid}">
        <findspec mode="all" disjoint="1">
          <conditions>
            <condition test="contains" flags="0">
              <category>
                <name internal="LcRevitData_Element">Element</name>
              </category>
              <property>
                <name internal="lcldrevit_parameter_-1002053">Workset</name>
              </property>
              <value>
                <data type="wstring">{workset}</data>
              </value>
            </condition>
            <condition test="contains" flags="0">
              <category>
                <name internal="LcOaNode">Item</name>
              </category>
              <property>
                <name internal="LcOaNodeLayer">Layer</name>
              </property>
              <value>
                <data type="wstring">{level}</data>
              </value>
            </condition>
          </conditions>
          <locator>/</locator>
        </findspec>
      </selectionset>
"""

# -----------------------------------
# Create Searchsets sheet (overwrite if exists)
# -----------------------------------
if "Searchsets" in wb.sheetnames:
    del wb["Searchsets"]

search_sheet = wb.create_sheet("Searchsets")
search_sheet.append(["Workset", "Level", "Searchset Name"])

# -----------------------------------
# Generate combinations
# -----------------------------------
for ws in worksets:
    for lvl in levels:
        selection_name = f"{ws} | {lvl}"

        output += selectionset_template.format(
            name=selection_name,
            guid=generate_guid(),
            workset=ws,
            level=lvl
        )

        search_sheet.append([ws, lvl, selection_name])

# -----------------------------------
# XML Footer
# -----------------------------------
output += """    </viewfolder>
  </selectionsets>
</exchange>"""

# -----------------------------------
# Save XML
# -----------------------------------
with open(xml_output_path, "w", encoding="utf-8") as f:
    f.write(output)

# -----------------------------------
# Save Excel
# -----------------------------------
wb.save(excel_path)

print("\nXML exported to:", xml_output_path)
print("Searchsets sheet created successfully.")
```

```python
import uuid
import os
from openpyxl import load_workbook

# -----------------------------------
# Ask user for folder path
# -----------------------------------
folder_path = input("Enter folder path containing Excel files: ").strip()

if not os.path.isdir(folder_path):
    print("Folder not found.")
    exit()

# -----------------------------------
# GUID Generator
# -----------------------------------
def generate_guid():
    return str(uuid.uuid4())

# -----------------------------------
# Process each Excel file
# -----------------------------------
for file_name in os.listdir(folder_path):

    if file_name.endswith(".xlsx") and not file_name.startswith("~$"):
        excel_path = os.path.join(folder_path, file_name)
        print(f"\nProcessing: {file_name}")

        try:
            wb = load_workbook(excel_path)

            # Check required sheets
            if "Worksets" not in wb.sheetnames or "Levels" not in wb.sheetnames:
                print("  Skipped (Missing Worksets or Levels sheet)")
                continue

            ws_sheet = wb["Worksets"]
            lvl_sheet = wb["Levels"]

            # Read data (skip header)
            worksets = [r[0].value for r in ws_sheet.iter_rows(min_row=2) if r[0].value]
            levels = [r[0].value for r in lvl_sheet.iter_rows(min_row=2) if r[0].value]

            # -----------------------------------
            # XML Header
            # -----------------------------------
            output = f"""<?xml version="1.0" encoding="UTF-8" ?>
<exchange xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance"
xsi:noNamespaceSchemaLocation="http://download.autodesk.com/us/navisworks/schemas/nw-exchange-12.0.xsd"
units="ft" filename="" filepath="">
  <selectionsets>
    <viewfolder name="WORKSETS_LEVELS" guid="{generate_guid()}">
"""

            # -----------------------------------
            # Selectionset Template
            # -----------------------------------
            selectionset_template = """      <selectionset name="{name}" guid="{guid}">
              <findspec mode="all" disjoint="1">
                <conditions>
                  <condition test="contains" flags="0">
                    <category>
                      <name internal="LcRevitData_Element">Element</name>
                    </category>
                    <property>
                      <name internal="lcldrevit_parameter_-1002053">Workset</name>
                    </property>
                    <value>
                      <data type="wstring">{workset}</data>
                    </value>
                  </condition>
                  <condition test="contains" flags="0">
                    <category>
                      <name internal="LcOaNode">Item</name>
                    </category>
                    <property>
                      <name internal="LcOaNodeLayer">Layer</name>
                    </property>
                    <value>
                      <data type="wstring">{level}</data>
                    </value>
                  </condition>
                </conditions>
                <locator>/</locator>
              </findspec>
            </selectionset>
      """


            # -----------------------------------
            # Create / Overwrite Searchsets sheet
            # -----------------------------------
            if "Searchsets" in wb.sheetnames:
                del wb["Searchsets"]

            search_sheet = wb.create_sheet("Searchsets")
            search_sheet.append(["Workset", "Level", "Searchset Name"])

            # -----------------------------------
            # Generate combinations
            # -----------------------------------
            for ws in worksets:
                for lvl in levels:
                    selection_name = f"{ws} | {lvl}"

                    output += selectionset_template.format(
                        name=selection_name,
                        guid=generate_guid(),
                        workset=ws,
                        level=lvl
                    )

                    search_sheet.append([ws, lvl, selection_name])

            # -----------------------------------
            # XML Footer
            # -----------------------------------
            output += """    </viewfolder>
  </selectionsets>
</exchange>"""

            # -----------------------------------
            # Save XML (same name as Excel)
            # -----------------------------------
            base_name = os.path.splitext(file_name)[0]
            xml_output_path = os.path.join(folder_path, base_name + ".xml")

            with open(xml_output_path, "w", encoding="utf-8") as f:
                f.write(output)

            # Save updated Excel
            wb.save(excel_path)

            print(f"  XML created: {base_name}.xml")
            print("  Searchsets sheet updated.")

        except Exception as e:
            print(f"  Error processing {file_name}: {e}")

print("\nBatch process completed.")
```
